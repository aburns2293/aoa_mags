```python
import py3Dmol
from colabfold.colabfold import plot_plddt_legend
import base64

hydrophobic_residues = ["GLY", "ALA", "VAL", "LEU", "ILE", "MET", "PHE", "TRP", "PRO"]
hydrophilic_residues = ["SER", "THR", "ASN", "GLN", "CYS", "SEC", "TYR"]

color_map = {}
for res in hydrophobic_residues: color_map[res] = 'blue'
for res in hydrophilic_residues:      color_map[res] = 'green'

pdb_file="ColabFold/EC11NC_v1_3248_ColabFold/EC11NC_v1_3248_ff802_unrelaxed_rank_001_alphafold2_ptm_model_5_seed_000.pdb"
view = py3Dmol.view(js='https://3dmol.org/build/3Dmol.js')
view.addModel(open(pdb_file, 'r').read(), 'pdb')
view.setStyle({'cartoon': {'color': 'gray'}})

    
model = view.getModel()
print(model)

    # Apply the custom colors based on the map
for res_name, color in color_map.items():
    view.addStyle({'resn': res_name}, {'cartoon': {'color': color}})

view.zoomTo()
```

```python
import base64

view.render()

png_data = view.png()

if "," in png_data:
    base64_string=png_data.split(",")[1]
else:
    base64_string=png_data

binary_data=base64.b64decode(base64_string)

with open('EC11NC_v1_2843_out.png', 'wb') as f:
    f.write(binary_data)
```

```python
# extract phasin hits

import pandas as pd
from Bio import SeqIO

df = pd.read_csv("AOA_genomes_phasin_hits.out", sep="\t", header=None, names=["query", "subject", "identity", "alignment_length", "mismatches", "gap_opens", "q_start", "q_end", "s_start", "s_end", "evalue", "bit_score"])

df_filtered = df[df['evalue'] < 1e-10]
df_first_hits = df_filtered.sort_values('evalue').groupby('query').first()

with open("AOA_genomes_phasin_hits.fasta", "w") as output_handle:
   for record in SeqIO.parse("all_AOA_proteins.fasta", "fasta"):
        if record.id in df_first_hits.index:
            SeqIO.write(record, output_handle, "fasta")
```
